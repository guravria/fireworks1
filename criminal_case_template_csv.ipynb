```python
import pandas as pd
import os

# STEP 1: Automatically create a starter template CSV if it doesn't exist
csv_filename = "criminal_case_template.csv"

if not os.path.exists(csv_filename):
    template_data = {
        "Witness_ID": ["W-01", "W-02", "W-03"],
        "Witness_Name": ["Ramesh Shinde", "Dr. Anita Joshi", "Sunil Gawde"],
        "Witness_Type": ["Eyewitness", "Medical Officer", "Panchnama Witness"],
        "Police_Statement_Summary": [
            "Saw accused running away at 9:30 PM with an iron rod.",
            "Injuries caused by blunt object; time of death 9 PM - 10 PM.",
            "Present during recovery of mobile phone from accused's house."
        ],
        "Court_Deposition_Notes": [
            "Cross-examination: Admitted it was dark and fog was heavy; couldn't clearly see face.",
            "Confirmed weapon dimensions match the wound pattern.",
            "Hostile turn: Denied police recovered phone in his presence."
        ],
        "Contradiction_Flag": ["HIGH - Visibility issue", "None", "CRITICAL - Witness turned hostile"]
    }
    df_init = pd.DataFrame(template_data)
    df_init.to_csv(csv_filename, index=False)
    print(f"✅ Created starter template file: {csv_filename}\n")

# STEP 2: Load the case file data
print(f"📂 Loading case file: {csv_filename}...\n")
df_cases = pd.read_csv(csv_filename)

# STEP 3: Display the main interactive table
print("=" * 60)
print("🚨 MASTER WITNESS & CONTRADICTION INDEX")
print("=" * 60)
display(df_cases)

# STEP 4: Automatically filter for critical loopholes / contradictions
print("\n" + "=" * 60)
print("🔍 AUTOMATICALLY FLAGGED FLAWS & CONTRADICTIONS:")
print("=" * 60)

if "Contradiction_Flag" in df_cases.columns:
    critical_loopholes = df_cases[df_cases["Contradiction_Flag"].str.contains("HIGH|CRITICAL|Hostile", case=False, na=False)]

    if not critical_loopholes.empty:
        display(critical_loopholes[["Witness_ID", "Witness_Name", "Witness_Type", "Contradiction_Flag"]])
    else:
        print("No critical contradictions flagged yet in this file.")
```
